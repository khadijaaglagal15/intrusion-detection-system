```python
import pandas as pd

# Charger le dataset
file_path = "dataset/Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv"
df = pd.read_csv(file_path)

# Nettoyer les noms des colonnes
df.columns = df.columns.str.strip()

# Nettoyer le nom du label
df["Label"] = df["Label"].str.strip()

# Sélectionner uniquement les attaques
attack_df = df[df["Label"].str.contains("Web Attack", case=False, na=False)].copy()

print("Nombre d'attaques :", len(attack_df))
print("\nTypes d'attaques :")
print(attack_df["Label"].value_counts())
```

```python
# Créer un fichier de test avec uniquement des attaques
attack_test = attack_df.sample(
    n=500,
    replace=True,
    random_state=42
)

# Les 23 features utilisées par le modèle
features = [
    "Flow Duration",
    "Total Length of Fwd Packets",
    "Fwd Packet Length Max",
    "Fwd Packet Length Mean",
    "Fwd Packet Length Std",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Flow IAT Std",
    "Flow IAT Max",
    "Fwd IAT Max",
    "Fwd Header Length",
    "Fwd Packets/s",
    "Bwd Packets/s",
    "Max Packet Length",
    "Packet Length Mean",
    "Packet Length Std",
    "Packet Length Variance",
    "Average Packet Size",
    "Avg Fwd Segment Size",
    "Fwd Header Length.1",
    "Subflow Fwd Bytes",
    "Init_Win_bytes_forward",
    "Init_Win_bytes_backward"
]

# Garder uniquement les features nécessaires
attack_test = attack_test[features]

# Supprimer les valeurs manquantes
attack_test = attack_test.dropna()

# Sauvegarder
attack_test.to_csv("attack_test.csv", index=False)

print("\nFichier créé : attack_test.csv")
print("Nombre de lignes :", len(attack_test))
```
